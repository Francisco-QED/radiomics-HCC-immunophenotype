```python
import os
import pydicom
import pandas as pd
import numpy as np
from collections import Counter

#script to pull basic demographics data from DICOM header for AI-Crohns project

startDir = '/Volumes/T7/regeneron_radiomics/filtered_DICOMS_april_2025'
key = pd.read_excel('/Users/Francisco/Desktop/RestrepoProjects/Regeneron retrospective radiomics/databases/Final_patient_key.xlsx')
#startDir = pd.read_excel('/Users/Francisco/Desktop/RestrepoProjects/AI-Crohns/data_sheets/segmented_cases_dec2024.xlsx', sheet_name = 'Sheet2' )
patients = [x for x in os.listdir(startDir) if x[0] != '.']


metaData = pd.DataFrame(columns = list(['Code','Age','Weight','Sex', 'Manufacturer','Magnetic Field Strength','Date of MRI'])) 

for patient in os.listdir(startDir):
    if os.path.isdir(os.path.join(startDir,patient)) == True:
                 
        series_list = [x for x in (os.listdir(os.path.join(startDir,patient))) if x[0]!='.']
        series = series_list[0]
        images = os.listdir(os.path.join(startDir,patient,series))
        image = images[0]
        imagePath = os.path.join(startDir,patient,series,image)                
        ds = pydicom.filereader.dcmread(imagePath)
        MRN = ds.PatientID
        name = ds.PatientName
        age = ds.PatientAge
        age = int(age[1:3])
        weight = round(float(ds.PatientWeight),2)
        sex = ds.PatientSex
        date = ds.StudyDate
        scanner = ds.Manufacturer
        model = ds.ManufacturerModelName
        thickness = ds.SliceThickness
        series = ds.SeriesDescription
        field = ds.MagneticFieldStrength
        new_rows = [{'Code': MRN, 'Age': age, 'Weight': weight, 'Sex': sex,  'Manufacturer':model,'Magnetic Field Strength':field, 'Date of MRI' : date}]         
        new_stuff = pd.DataFrame(new_rows)
        metaData = pd.concat([metaData, new_stuff], ignore_index = True)
        #print(os.path.join(startDir,patient,studies))
    

#mrns = metaData['MRN']

# Find MRNs from key file

metaData = metaData.merge(key[['Code', 'MRN']], on='Code', how='left')

col_order = ['MRN'] + [c for c in metaData.columns if c != 'MRN']
metaData = metaData[col_order]

metaData1=metaData.dropna(subset = 'MRN')


startDir = '/Volumes/LaCie2/SA2_DICOMS/Filtered_pre-treatment'

metaData = pd.DataFrame(columns = list(['Code','Age','Weight','Sex', 'Manufacturer','Magnetic Field Strength','Date of MRI'])) 

for patient in os.listdir(startDir):
    if os.path.isdir(os.path.join(startDir,patient)) == True:
                 
        series_list = [x for x in (os.listdir(os.path.join(startDir,patient))) if x[0]!='.']
        series = series_list[0]
        images = os.listdir(os.path.join(startDir,patient,series))
        image = images[0]
        imagePath = os.path.join(startDir,patient,series,image)                
        ds = pydicom.filereader.dcmread(imagePath)
        MRN = ds.PatientID
        name = ds.PatientName
        age = ds.PatientAge
        age = int(age[1:3])
        weight = round(float(ds.PatientWeight),2)
        sex = ds.PatientSex
        date = ds.StudyDate
        scanner = ds.Manufacturer
        model = ds.ManufacturerModelName
        thickness = ds.SliceThickness
        series = ds.SeriesDescription
        field = ds.MagneticFieldStrength
        new_rows = [{'Code': MRN, 'Age': age, 'Weight': weight, 'Sex': sex,  'Manufacturer':model,'Magnetic Field Strength':field, 'Date of MRI' : date}]         
        new_stuff = pd.DataFrame(new_rows)
        metaData = pd.concat([metaData, new_stuff], ignore_index = True)
        #print(os.path.join(startDir,patient,studies))
    #break

#mrns = metaData['MRN']

# Find MRNs from key file

metaData = metaData.merge(key[['Code', 'MRN']], on='Code', how='left')

col_order = ['MRN'] + [c for c in metaData.columns if c != 'MRN']
metaData = metaData[col_order]

#metaData = metaData.drop('Code', axis=1)

metaData2=metaData.dropna(subset = 'MRN')

metaData = pd.concat([metaData1,metaData2])

metaData
```
