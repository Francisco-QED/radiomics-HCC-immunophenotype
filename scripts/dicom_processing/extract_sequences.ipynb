```python
import os
import pandas as pd
import shutil
import pydicom
import numpy as np
import zipfile
```

```python
#########   Sinai   #########

# Full MRI folders come in a similar structure as those for UW, so the script is fairly similar

dicomDir = '/Volumes/T7/regeneron_radiomics/DICOMS'
outDir = '/Volumes/T7/regeneron_radiomics/filtered_DICOMS'
```

```python
keys = ['haste','ssfs','ss-fse','sstse','ssh-tse','ufse','ax t2','ax 80 te','lava','fame','vibe','thrive','Ax fspgr Hepatobiliary',
        'dwi','adc','diffusion','ep2d_diff']
exclude = ['dixon', 'dce', 'cor','localizer','inphase','outphase','exponential','fatsat','t2*','sub', '_filt_','T2-star','radial'] 
patients = sorted([x for x in os.listdir(dicomDir) if x[0] != '.'])

for patient in patients:
    studies = sorted([x for x in os.listdir(os.path.join(dicomDir,patient)) if x[0] != '.'])
    for study in studies:
        seriess = sorted([x for x in os.listdir(os.path.join(dicomDir,patient,study)) if x[0] != '.'])
                                
        for series in seriess:        
            if any(elem in series.casefold() for elem in keys) and not any(elem in series.casefold() for elem in exclude): 
                series_path = os.path.join(dicomDir,patient,study,series)
                images = os.listdir(series_path)
                image = images[0]
                image_path = os.path.join(series_path,image) 
             
                ds = pydicom.filereader.dcmread(image_path)
                series_time = ds.SeriesTime
                series_time = series_time.split('.')[0]
                new_series_name = series + '_' + series_time          # maybe later we'd like to include series time in name
                dest_path = os.path.join(outDir,patient,series)
                
                if os.path.exists(dest_path) == False:
                    shutil.copytree(series_path,dest_path)
                    
                 
```
