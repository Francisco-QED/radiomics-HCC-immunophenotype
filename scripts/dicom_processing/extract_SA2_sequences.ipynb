```python
# B1/B2 DICOMS for Anis to easily segment. These came zipped by Marco, so I'll have to unzip them first

import zipfile
import os

keys = ['ax haste t2','lava','vibe part 1','1 min','3 min', '20 min','pre-caipi',
        'dwi','adc','diffusion','ep2d_diff','tra_bh_f','tra_bh_ff','tra_bh_w','tra_bh_wf']
exclude = ['scout','_sub'] 

extract_to = '/Volumes/LaCie2/SA2_DICOMS/Filtered_post-treatment'
rootDir = '/Volumes/LaCie2/SA2_DICOMS/Post-treatment_deid'
zip_patients = sorted([x for x in os.listdir(rootDir) if x[0] != '.'])

for zip_patient in zip_patients:
    zip_patient_path = os.path.join(rootDir,zip_patient)
    with zipfile.ZipFile(zip_patient_path, 'r') as zf:
        for member in zf.namelist():
            if any(elem in member.casefold() for elem in keys) and not any(elem in member.casefold() for elem in exclude):
                print(f"Extracting: {member}")
                zf.extract(member, extract_to)
   
```
