```python
import os
import pandas as pd
import shutil
import pydicom
```

```python

# Script to extract the T1 pre-con and T1-AP phases from a DCE-MRI acquisition. 

dicomDir = '/Volumes/LaCie2/SA2_DICOMS/Pre-treatment_deid/CE2-2301-021/RESEARCH PROTOCOLS_TAOULI_MSSM^REGENERON/AX DCE VIBE PART 1'
outDir = '/Volumes/LaCie2/SA2_DICOMS/Pre-treatment_deid/CE2-2301-021'

images = sorted([x for x in os.listdir(dicomDir) if x[0] != '.'])

# For patient 81, I know there are 10 timepoints and 160 images

numTvals = 16
numFiles = len(images)

filesTval = numFiles/numTvals
filesTval


```

```python

# Now, I manually pick the time point corressponding to the best AP

t_coord = 9   # Acquisition number

for image in images:
    image_path = os.path.join(dicomDir,image)
    instance_number = pydicom.dcmread(image_path).InstanceNumber

    if instance_number>(t_coord-1)*filesTval and instance_number<(t_coord-1)*filesTval+45:
        series = 'T1 VIBE AP2'
        new_image_path = os.path.join(outDir,series,image)
        if os.path.exists(new_image_path) == False:
            shutil.copyfile(image_path,new_image_path)

```
