```python
import os
import pydicom
import pandas as pd
import numpy as np
from collections import Counter


#startDir = '/Users/taoulilab/Desktop/Restrepo_projects/FAST MRI HCC/Sinai/DICOMS'

startDir = '/Users/Francisco/Desktop/T-LAB/T-LAB/DICOM ANONYMOUS/Regeneron retrospective radiomics/DICOMS_11_28_24'

metaData = pd.DataFrame(columns = list(['Name','MRN','Manufacturer','Study Date','Accession','Contrast'])) 
patients = [os.path.join(startDir,patient) for patient in os.listdir(startDir) if patient[0] != '.']



for patient in patients:
    i=0
    
    mri_studies = [os.path.join(patient,study) for study in os.listdir(patient) if study[0] != '.']
    
    for mri_study in mri_studies:

        mri_study_path = os.path.join(patient,mri_study)
        mri_seriess = [os.path.join(mri_study,series) for series in os.listdir(mri_study_path) if series[0] != '.']


        for mri_series in mri_seriess:
                           
            if 'MIN' in os.path.split(mri_series)[1] or 'POST' in os.path.split(mri_series)[1] or 'Min' in os.path.split(mri_series)[1] or 'Post' in os.path.split(mri_series)[1] :
                i = 1
                
                mri_series_path = os.path.join(mri_study_path,mri_series)
                mri_images = [x for x in os.listdir(mri_series_path) if x[0]!='.']
                mri_image = mri_images[0]
                mri_image_path = os.path.join(mri_series_path,mri_image) 
                #print(mri_image_path)
                ds = pydicom.filereader.dcmread(mri_image_path)
                name = ds.PatientName
                MRN = ds.PatientID
                manufacturer = ds.Manufacturer
                date=ds.StudyDate
                accession = ds.AccessionNumber
                
                if 'ContrastBolusAgent' in ds:
                    contrast = ds.ContrastBolusAgent
                else:
                    contrast = 'no contrast'
                    
                new_rows = [{'Name': str(name), 'MRN': MRN, 'Manufacturer':manufacturer,'Study Date':date,'Accession':accession, 'Contrast':contrast}]         
                new_stuff = pd.DataFrame(new_rows)
                metaData = pd.concat([metaData, new_stuff], ignore_index = True)
                break
                
        if i==0:
            mri_series_path = os.path.join(mri_study_path,mri_series)
            mri_images = [x for x in os.listdir(mri_series_path) if x[0]!='.']
            mri_image = mri_images[0]
            mri_image_path = os.path.join(mri_series_path,mri_image) 
            #print(mri_image_path)
            ds = pydicom.filereader.dcmread(mri_image_path)
            name = ds.PatientName
            MRN = ds.PatientID
            manufacturer = ds.Manufacturer
            date=ds.StudyDate
            accession = ds.AccessionNumber
            contrast = 'no contrast'
            new_rows = [{'Name': str(name), 'MRN': MRN, 'Manufacturer':manufacturer, 'Contrast':contrast}]         
            new_stuff = pd.DataFrame(new_rows)
            metaData = pd.concat([metaData, new_stuff], ignore_index = True)
            
    #print(patient,i)
metaData         

```
