```python
import os
import pandas as pd
import shutil
import pydicom

# Script to clean up unruly DWI sequences with b-values lumped to gether and not separated by Slicer

dicomDir = '/Users/Francisco/Downloads/Loehmann_Karl/Ra_Mri_Prostate_Without_Coil_With_And_Without_Contrast - 1008412611/DWI_b50_1000_1600_CALC2000_12921_TRACEW_4'
outDir = '/Users/Francisco/Downloads'

images = sorted([x for x in os.listdir(dicomDir) if x[0] != '.'])

# For patient 131, I know there are 3 b-values and 114 images

numBvals = 3
numFiles = len(images)

filesBval = numFiles/numBvals
filesBval
```

```python
for image in images:
    image_path = os.path.join(dicomDir,image)
    instance_number = pydicom.dcmread(image_path).InstanceNumber

    if instance_number<filesBval+1:
        series = 'b50'
        new_image_path = os.path.join(outDir,series,image)
        if os.path.exists(new_image_path) == False:
            shutil.copyfile(image_path,new_image_path)

    elif instance_number>filesBval and instance_number<2*filesBval+1:
        series = 'b1000'
        new_image_path = os.path.join(outDir,series,image)
        if os.path.exists(new_image_path) == False:
            shutil.copyfile(image_path,new_image_path)

    elif instance_number>2*filesBval and instance_number<3*filesBval+1:
        series = 'b1600'
        new_image_path = os.path.join(outDir,series,image)
        if os.path.exists(new_image_path) == False:
            shutil.copyfile(image_path,new_image_path)

    #elif instance_number>3*filesBval and instance_number<4*filesBval+1:
     #   series = 'Ax eDWI_b800'
      
        new_image_path = os.path.join(outDir,series,image)
       
        if os.path.exists(new_image_path) == False:
            shutil.copyfile(image_path,new_image_path)

  
```
